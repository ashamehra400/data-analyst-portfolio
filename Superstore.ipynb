```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.columns
```

# Bar chart-Sales by Category

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

This chart helps identify which category generates the most revenue.
The category with the highest bar is the main revenue driver, while lower bars show categories that may need improvement or promotion.

## Line Chart -Sales Trend Over Time

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

monthly_sales.plot(kind="line")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

This chart helps identify growth trends, seasonal patterns, and periods of high or low demand, which supports forecasting and planning decisions.

# Pie Chart-Sales Share by Category

```python
category_sales.plot(kind="pie", autopct="%1.1f%%")
plt.title("Sales Share by Category")
plt.ylabel("")
plt.show()
```

This chart makes it easy to understand how much each category contributes to overall sales and which categories dominate the business.

# Bar Chart – Profit by Sub-Category

```python
sub_profit = df.groupby("Sub-Category")["Profit"].sum()

sub_profit.plot(kind="bar")
plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")
plt.show()
```

This chart helps identify which sub-categories are profitable and which ones are causing losses, enabling better pricing and cost-control decisions.
